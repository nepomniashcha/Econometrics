```python
# критерій Дарбіна-Уотсона
# Крок 1: Знаходимо розрахункові (теоретичні) значення Y_pred
df['Y_pred'] = (
    coefficients['X0'] * df['X0'] + 
    coefficients['X1'] * df['X1'] + 
    coefficients['X2'] * df['X2'] + 
    coefficients['X3'] * df['X3'] + 
    coefficients['X4'] * df['X4'] + 
    coefficients['X5'] * df['X5']
)

new_df = pd.DataFrame()
# Крок 2: Знаходимо залишки моделі (u_i)
# Для цього віднімаємо прогноз від фактичних значень
new_df['u'] = df['Y1'] - df['Y_pred']

# Крок 3: Обчислюємо чисельник (сума квадратів різниць послідовних залишків)
# Використовуємо .diff() для знаходження (u_i - u_{i-1})
new_df['u_diff'] = new_df['u'].diff() 
numerator = (new_df['u_diff'] ** 2).sum()

# Крок 4: Обчислюємо знаменник (сума квадратів залишків)
denominator = (new_df['u'] ** 2).sum()

# Крок 5: Знаходимо статистику Дарбіна-Уотсона
d_ct = numerator / denominator

print(f"Чисельник: {numerator:.4f}")
print(f"Знаменник: {denominator:.4f}")
print(f"Статистика Дарбіна-Уотсона (d_CT): {d_ct:.4f}")
```

```python
def check_durbin_watson_hypothesis(d_ct, d_0, d_u):
    """
    Перевіряє гіпотезу про наявність автокореляції на основі 
    розрахованої статистики d_ct та критичних значень d_0, d_u.
    """
    if d_ct < 0 or d_ct > 4:
        return "Помилка: d_ct має бути в діапазоні [0; 4]."
        
    print(f"Розраховане значення d_CT: {d_ct:.4f}")
    print(f"Критичні межі: d_0 = {d_0}, d_u = {d_u}\n")

    # 1. Зона відхилення H0 (позитивна автокореляція)
    if 0 <= d_ct <= d_0:
        return "Результат (—): Відхилення H0. Присутня додатна автокореляція залишків."
        
    # 2. Зона невизначеності (ближче до нуля)
    elif d_0 < d_ct <= d_u:
        return "Результат (?): Область невизначеності. Згідно з правилом алгоритму, приймається гіпотеза про наявність автокореляції."
        
    # 3. Зона прийняття H0 (відсутність автокореляції)
    elif d_u < d_ct < (4 - d_u):
        return "Результат (+): Прийняття H0. Відсутня автокореляція залишків."
        
    # 4. Зона невизначеності (ближче до 4)
    elif (4 - d_u) <= d_ct < (4 - d_0):
        return "Результат (?): Область невизначеності. Згідно з правилом алгоритму, приймається гіпотеза про наявність автокореляції."
        
    # 5. Зона відхилення H0 (негативна автокореляція)
    elif (4 - d_0) <= d_ct <= 4:
        return "Результат (—): Відхилення H0. Присутня від'ємна автокореляція залишків."


# d_0 та d_u визначаються за таблицями Дарбіна-Уотсона 
# залежно від рівня значущості (alpha), кількості спостережень (n) 
# та кількості пояснюючих змінних (k).

# Припустимо, ми знайшли такі табличні значення для нашої вибірки:
d_0_table = 1.038
d_u_table = 1.767


d_ct_calculated = 2.0336 
print(check_durbin_watson_hypothesis(d_ct_calculated, d_0_table, d_u_table))
```

```python
from statsmodels.stats.stattools import durbin_watson

def cochrane_orcutt(X, y, max_iter=100, tol=1e-5):
    """
    Реализация метода Кохрейна-Оркатта для устранения автокорреляции остатков.
    
    Параметры:
    X        : независимые переменные (numpy array или pandas DataFrame)
    y        : зависимая переменная (numpy array или pandas Series)
    max_iter : максимальное количество итераций
    tol      : порог сходимости коэффициентов
    """
    # Гарантируем работу с numpy arrays
    X = np.asarray(X)
    y = np.asarray(y)
    n_obs = len(y)
    
    # Добавляем константу, если её нет
    if not np.any(np.all(X == 1, axis=0)):
        X = sm.add_constant(X)
        
    beta_prev = np.zeros(X.shape[1])
    beta = sm.OLS(y, X).fit().params # Начальная 1МНК оценка
    
    for iteration in range(max_iter):
        # 1) Рассчитываем остатки u_t
        u = y - X @ beta
        
        # 2) Застосовуємо тест Дарбіна-Уотсона
        dw_stat = durbin_watson(u)
        
        # Условие выхода (сходимость коэффициентов beta или отсутствие автокорреляции)
        if np.linalg.norm(beta - beta_prev) < tol:
            print(f"Алгоритм сошелся на итерации {iteration + 1}.")
            break
            
        beta_prev = beta.copy()
        
        # 3) Оцениваем параметр rho (с крышечкой)
        # Числитель: sum(u_t * u_{t-1}) со второго элемента
        # Знаменатель: sum(u_t^2) по всем элементам
        u_t = u[1:]
        u_t_1 = u[:-1]
        rho = np.sum(u_t * u_t_1) / np.sum(u**2)
        
        # 4) Преобразуем начальные данные с помощью матрицы преобразований T^A
        # Вместо умножения на огромную разреженную матрицу, применяем векторизованное преобразование:
        y_star = np.zeros(n_obs)
        X_star = np.zeros_like(X)
        
        # Трансформация первого наблюдения (первая строка матрицы T^A)
        y_star[0] = np.sqrt(1 - rho**2) * y[0]
        X_star[0] = np.sqrt(1 - rho**2) * X[0]
        
        # Трансформация остальных наблюдений (остальные строки матрицы T^A)
        y_star[1:] = y[1:] - rho * y[:-1]
        X_star[1:] = X[1:] - rho * X[:-1]
        
        # 5) По новым данным находим 1МНК-оценки для параметра beta
        model_star = sm.OLS(y_star, X_star)
        results_star = model_star.fit()
        beta = results_star.params
        
        # 6) Переход к п.1 происходит автоматически за счет цикла for
        
    print(f"Финальная статистика Дарбина-Уотсона: {dw_stat:.4f}")
    print(f"Оценка параметра rho: {rho:.4f}")
    
    # Возвращаем финальную модель и полученные параметры
    return results_star, rho, dw_stat



# 1. Виокремлюємо незалежні змінні (матрицю X) з вашого df
X_columns = ['X0', 'X1', 'X2', 'X3', 'X4', 'X5']
X_data = df[X_columns]

# 2. Виокремлюємо залежну змінну (вектор Y)
Y_data = df['Y1']

# (Опціонально) Якщо вам потрібно обчислити початкові залишки u_error саме за вашими коефіцієнтами:
Y_pred = sum(df[col] * coefficients[col] for col in X_columns)
u_error = Y_data - Y_pred

# 3. Запуск алгоритму Кохрейна-Оркатта на ваших РЕАЛЬНИХ даних
# (Використовуємо функцію cochrane_orcutt з попереднього кроку)
final_model, estimated_rho, final_dw = cochrane_orcutt(X_data, Y_data)

print("\n--- Фінальні коефіцієнти моделі після усунення автокореляції ---")
print(final_model.summary().tables[1])
```
