```python
models_results = {}
```

```python
# Визначення залежної змінної (Y) та незалежних факторів (X)
y = df['Y1']
X = df[['X1', 'X2', 'X3', 'X4', 'X5']]

# Додавання константи (вільного члена) до моделі
# Це забезпечує виконання умови У1 щодо відсутності систематичних помилок
X = sm.add_constant(X)

# Побудова моделі 1МНК (Ordinary Least Squares - OLS)
model_A = sm.OLS(y, X)

# Оцінювання параметрів моделі
results_A = model_A.fit()

models_results['A'] = results_A

# Виведення результатів
print(results_A.summary())
```

```python
# Перетворення у матриці numpy
y_mat = df['Y1'].values
# Додаємо стовпець одиниць для константи
X_mat = np.column_stack((np.ones(len(df)), df[['X1', 'X2', 'X3', 'X4', 'X5']].values))

# Розрахунок бета за формулою: (X^T * X)^(-1) * X^T * Y
beta_hat = np.linalg.inv(X_mat.T @ X_mat) @ X_mat.T @ y_mat

print("Оцінені параметри бета для моделі А:", beta_hat)
```

```python
y = df['Y1']
factors = ['X1', 'X2', 'X3', 'X4', 'X5']
model_names = ['Б', 'В', 'Г', 'Д', 'Е']

# Спосіб 1: Використання бібліотеки statsmodels
print("--- Результати через statsmodels ---")
for name, factor in zip(model_names, factors):
    X_single = df[factor]
    # Додаємо константу для включення вільного члена (beta1)
    X_single = sm.add_constant(X_single) 
    
    model = sm.OLS(y, X_single)
    results = model.fit()

    models_results[name] = results
    
    print(f"\nМодель {name} (Y від {factor}):")
    print(f"beta_1 (константа): {results.params['const']:.4f}")
    print(f"beta_2 (коефіцієнт): {results.params[factor]:.4f}")
    print(f"R-squared: {results.rsquared:.4f}")
```

```python
# Спосіб 2: Розрахунок за точними формулами з наданого зображення
print("\n--- Результати за аналітичними формулами ---")
def calculate_simple_regression(x, y):
    T = len(x)
    
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_xy = np.sum(x * y)
    sum_x_squared = np.sum(x**2)
    
    # Розрахунок коефіцієнта beta2_hat за формулою
    numerator = sum_xy - (1/T) * sum_x * sum_y
    denominator = sum_x_squared - (1/T) * (sum_x)**2
    beta2_hat = numerator / denominator
    
    # Середні значення
    y_bar = sum_y / T
    x_bar = sum_x / T
    
    # Розрахунок вільного члена beta1_hat
    beta1_hat = y_bar - beta2_hat * x_bar
    
    return beta1_hat, beta2_hat

for name, factor in zip(model_names, factors):
    b1, b2 = calculate_simple_regression(df[factor].values, y.values)
    # models_results[name] = [b1, b2]
    print(f"Модель {name} (Y від {factor}): beta1_hat = {b1:.4f}, beta2_hat = {b2:.4f}")
```

```python
def get_model_metrics(results, y_true):
    # u_hat = y_true - y_pred
    u_hat = results.resid 
    
    # 1. Дисперсия остатков (sigma^2_u)
    # Формула: сумма квадратов остатков / (T - N)[cite: 1]
    # sigma2_u = np.sum(u_hat**2) / (T - N)
    sigma2_u = results.scale 
    
    # 2. Mean Squared Error (MSE)
    # Формула: (1/T) * сумма квадратов остатков
    mse = np.mean(u_hat**2)
    
    # 3. Mean Absolute Percentage Error (MAPE)
    # Формула: (100% / T) * сумма (|остаток| / y_true)
    mape = np.mean(np.abs(u_hat / y_true)) * 100
    
    # 4. Математическое ожидание остатков M(u)
    m_u = np.mean(u_hat)
    
    # Формирование строки уравнения 
    eq_parts = []
    for var_name, coef in results.params.items():
        if var_name == 'const':
            # при матричных вычислениях часто вводят дополнительную фиктивную переменную, которая тождественно равна единице (x_0 = 1). 
            # Она нужна для того, чтобы рассчитать свободный член (константу) уравнения регрессии
            eq_parts.append(f"{coef:.3f} x_0")
        else:
            formatted_var = var_name.lower()
            if formatted_var.startswith('x') and '_' not in formatted_var:
                formatted_var = formatted_var.replace('x', 'x_')
                
            sign = '+' if coef >= 0 else '-'
            eq_parts.append(f"{sign} {abs(coef):.3g} {formatted_var}")
    
    equation = "y^ = " + " ".join(eq_parts).strip()
    
    if equation.startswith("y^ = + "):
        equation = equation.replace("y^ = + ", "y^ = ")
    
    return equation, sigma2_u, mse, mape, m_u

table_data = []

for model_name, res in models_results.items():
    eq, s2, mse, mape, mu = get_model_metrics(res, y)
    
    table_data.append({
        'Модель': model_name,
        'Рівняння': eq,
        'sigma^2_u': round(s2, 3),
        'MSE': round(mse, 3),
        'MAPE': f"{round(mape, 1)}%",
        'M(u)': f"{mu:.0e}".replace('e', ' * 10^') 
    })

df_summary = pd.DataFrame(table_data)
print(df_summary.to_string(index=False))
```
