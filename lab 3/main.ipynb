```python
# Допоміжні функції для розрахунку середнього та стандартного відхилення
def calc_mean(values):
    return sum(values) / len(values)

def calc_std(values, mean_val):
    T = len(values)
    variance = sum((v - mean_val) ** 2 for v in values) / (T - 1)
    return variance ** 0.5

# Назви колонок регресорів (переконайтеся, що ключі в coefficients такі самі)
x_cols = ['X1', 'X2', 'X3', 'X4', 'X5']

# 1. Розрахунок середніх значень та стандартних відхилень
# Використовуємо .tolist() для перетворення даних pandas у звичайні списки Python
means_X = {k: calc_mean(df[k].tolist()) for k in x_cols}
stds_X = {k: calc_std(df[k].tolist(), means_X[k]) for k in x_cols}

mean_Y = calc_mean(df['Y1'].tolist())
std_Y = calc_std(df['Y1'].tolist(), mean_Y)

print("--- Середні значення (Mean) ---")
for k in means_X:
    print(f"{k}: {means_X[k]:.4f}")
print(f"Y: {mean_Y:.4f}\n")

print("--- Середньоквадратичні відхилення (Std Dev) ---")
for k in stds_X:
    print(f"{k}: {stds_X[k]:.4f}")
print(f"X0: {0:.4f}\n")
print(f"Y: {std_Y:.4f}\n")

# 2. Розрахунок стандартизованих коефіцієнтів (бета)
standardized_betas = {}
for k in x_cols: # Проходим только по списку реальных переменных
    # Убедитесь, что ключи в словаре coefficients совпадают с x_cols
    standardized_betas[k] = coefficients[k] * (stds_X[k] / std_Y)

print("--- Стандартизовані коефіцієнти (Beta) ---")
for k, beta in standardized_betas.items():
    print(f"Beta_{k}: {beta:.6f}")
print(f"Beta_X0: {0:.4f}\n")
print()
```

```python
# 3. Загальний вигляд нової моделі у стандартизованому масштабі (t)
model_terms = [f"({beta:.4f})*t_{k}" for k, beta in standardized_betas.items()]
new_model = "t_y = " + " + ".join(model_terms)
print("--- Нова стандартизована модель ---")
print(new_model.replace("+ -", "- "))
print()

# 4. Визначення найбільш впливового регресора (найбільший за модулем коефіцієнт)
most_influential = max(standardized_betas, key=lambda k: abs(standardized_betas[k]))
max_beta_val = standardized_betas[most_influential]
print("--- Найбільш впливовий регресор ---")
print(f"Регресор: {most_influential} (Beta = {max_beta_val:.6f})")
```

```python
# Оцінена еластичність розраховується за формулою: (коефіцієнт * середнє_X) / середнє_Y
# Використовуємо словники means_X та mean_Y, розраховані у попередньому кроці
# Вільний член (x_0) виключається, оскільки еластичність розраховується для регресорів

elasticities = {}
for k in x_cols: # Проходимо лише по реальних змінних
    # Перевірка, щоб уникнути ділення на нуль
    if mean_Y != 0:
        elasticities[k] = coefficients[k] * (means_X[k] / mean_Y)
    else:
        elasticities[k] = None
        print("Помилка: середнє значення Y дорівнює нулю.")

print("--- Коефіцієнти еластичності ---")
for k, elast in elasticities.items():
    if elast is not None:
        print(f"E_{k}: {elast:.6f}")
```

```python
# Назви колонок регресорів
x_cols = ['X1', 'X2', 'X3', 'X4', 'X5']

n = len(df)
k_params = len(x_cols) + 1 # кількість параметрів (регресори + вільний член)

# Формуємо матрицю X (додаємо стовпець одиниць для вільного члена x_0)
# Використовуємо .values для перетворення pandas DataFrame у numpy array
X_matrix = np.column_stack((np.ones(n), df[x_cols].values))
Y_vector = df['Y1'].values

# 1. Розрахунок оцінених коефіцієнтів
# beta_hat = (X^T * X)^(-1) * X^T * Y
XTX = X_matrix.T @ X_matrix
XTX_inv = np.linalg.inv(XTX)
beta_hat = XTX_inv @ X_matrix.T @ Y_vector

# 2. Розрахунок дисперсії залишків (sigma^2_u)
Y_pred = X_matrix @ beta_hat
residuals = Y_vector - Y_pred

# Оцінка дисперсії залишків
sigma2_u = np.sum(residuals**2) / (n - k_params)

# 3. Дисперсійно-коваріаційна матриця оцінених коефіцієнтів
# Формула: Sigma_beta = sigma2_u * (X^T * X)^(-1)
cov_matrix_beta = sigma2_u * XTX_inv

print("--- Дисперсійно-коваріаційна матриця ---")
# Створимо DataFrame для красивого виводу
cols_labels = ['x_0'] + x_cols
df_cov = pd.DataFrame(cov_matrix_beta, index=cols_labels, columns=cols_labels)
print(df_cov.round(6))
print()

# 4. Кореляційна матриця оцінених коефіцієнтів
corr_matrix_beta = np.zeros((k_params, k_params))

for i in range(k_params):
    for j in range(k_params):
        cov_ij = cov_matrix_beta[i, j]
        std_i = np.sqrt(cov_matrix_beta[i, i])
        std_j = np.sqrt(cov_matrix_beta[j, j])
        
        # Формула кореляції: r = cov(i,j) / (std_i * std_j)
        corr_matrix_beta[i, j] = cov_ij / (std_i * std_j)

print("--- Кореляційна матриця оцінених коефіцієнтів ---")
df_corr = pd.DataFrame(corr_matrix_beta, index=cols_labels, columns=cols_labels)
print(df_corr.round(6))
```
