```python
import pandas as pd
import numpy as np
from scipy import stats

data = {
    'Країна': [
        'Австралія', 'Австрія', 'Аргентина', 'Бангладеш', 'Бельгія', 'Білорусь', 'Бразилія', 'Буркіна-Фасо', 
        'Велика Британія', "В'єтнам", 'Гаїті', 'Гондурас', 'Гонконг', 'Ефіопія', 'Єгипет', 'Замбія', 'Індія', 
        'Ірландія', 'Іспанія', 'Італія', 'Канада', 'Китай', 'Колумбія', 'Коста-Рика', 'Куба', 'Малайзія', 
        'Марокко', 'Мексика', 'Нідерланди', 'Німеччина', 'Нова Зеландія', 'Норвегія', 'ОАЕ', 'ПАР', 
        'Південна Корея', 'Північна Корея', 'Польща', 'Португалія', 'Росія', 'Саудівська Аравія', 'Сінгапур', 
        'США', 'Таїланд', 'Туреччина', 'Україна', 'Філіппіни', 'Фінляндія', 'Франція', 'Чилі', 'Швейцарія', 
        'Швеція', 'Японія'
    ],
    'Y1': [74, 73, 68, 53, 73, 66, 57, 47, 74, 63, 43, 65, 75, 51, 60, 44, 58, 73, 74, 74, 74, 67, 69, 76, 74, 66, 66, 69, 75, 73, 73, 74, 70, 62, 68, 67, 69, 71, 64, 66, 73, 73, 65, 69, 65, 63, 72, 74, 71, 75, 75, 76],
    'X1': [17800, 8000, 33900, 125000, 10100, 10300, 156600, 10000, 58400, 73100, 6500, 5600, 5800, 55200, 60000, 9100, 911600, 3600, 39200, 58100, 29100, 1205200, 35600, 3300, 11100, 19500, 28600, 91800, 15400, 81200, 3524, 4300, 2800, 43900, 45000, 23100, 38600, 10500, 149200, 18000, 2900, 260800, 59400, 62200, 51800, 69800, 5100, 58000, 14000, 7000, 8800, 125500],
    'X2': [15, 12, 20, 35, 12, 13, 21, 47, 13, 27, 40, 35, 13, 45, 29, 46, 29, 14, 11, 11, 14, 21, 24, 26, 17, 29, 29, 28, 13, 11, 16, 13, 28, 34, 16, 24, 14, 12, 13, 38, 16, 15, 19, 26, 12, 27, 13, 13, 23, 12, 14, 11],
    'X3': [8, 11, 9, 11, 11, 11, 9, 18, 11, 8, 19, 6, 6, 14, 9, 18, 10, 9, 9, 10, 8, 7, 6, 4, 7, 5, 6, 5, 9, 11, 8, 10, 3, 8, 6, 6, 10, 10, 11, 6, 6, 9, 6, 6, 13, 7, 10, 9, 6, 9, 11, 7],
    'X6': [16848, 18396, 3408, 202, 17912, 6500, 2354, 357, 15974, 230, 383, 1030, 14641, 122, 748, 573, 275, 12170, 13047, 17500, 19904, 377, 1538, 2031, 1382, 2995, 1062, 3604, 17245, 17539, 14381, 17755, 14193, 3128, 6627, 1000, 4429, 9000, 6680, 6651, 14990, 23474, 1800, 3721, 2340, 867, 15877, 18944, 2591, 22384, 16900, 19860],
    'X8': [85, 58, 86, 16, 96, 65, 75, 15, 89, 20, 29, 44, 94, 12, 44, 42, 26, 57, 78, 69, 77, 26, 70, 47, 74, 43, 46, 73, 89, 85, 84, 75, 81, 49, 72, 60, 62, 34, 74, 77, 100, 75, 22, 61, 67, 43, 60, 73, 85, 62, 84, 77]
}
df = pd.DataFrame(data)
```

```python
cols = ['X1', 'X2', 'X3', 'X6', 'X8']
X = df[cols].dropna()

T, N = X.shape          # T - кількість спостережень, N - кількість змінних (5)
alpha = 0.05            # рівень значущості (ймовірність 95%)

# метод Фаррара-Глобера
X_norm = (X - X.mean()) / X.std(ddof=1)

# ==============================================================================
# 1. КОРЕЛЯЦІЙНА МАТРИЦЯ r
# ==============================================================================
r = X_norm.corr().values
r_df = pd.DataFrame(r, index=cols, columns=cols)
print("1. Кореляційна матриця r:")
print(r_df.round(4), "\n")

# ==============================================================================
# 2. ПЕРЕВІРКА НАЯВНОСТІ МУЛЬТИКОЛІНЕАРНОСТІ У ВСЬОМУ МАСИВІ (Критерій Хі-квадрат)
# ==============================================================================
det_r = np.linalg.det(r)
chi2_stat = -(T - 1 - (2 * N + 5) / 6) * np.log(det_r)
df_chi2 = int(N * (N - 1) / 2)
chi2_crit = stats.chi2.ppf(1 - alpha, df_chi2)

has_chi2 = chi2_stat > chi2_crit
print(f"2. Загальна мультиколінеарність (χ²-критерій):")
print(f"   det(r) = {det_r:.6f}")
print(f"   χ²_СТ = {chi2_stat:.4f}, χ²_КР(0.05; {df_chi2}) = {chi2_crit:.4f}")
print(f"   Висновок: {f'З ймовірністю {(1 - alpha)*100:.0f}% у масиві незалежних змінних існує мультиколінеарність' if has_chi2 else 'Мультиколінеарність відсутня'}\n")
```

```python
# ==============================================================================
# 3. ОБЕРНЕНА МАТРИЦЯ Z = r^(-1) ТА F-КРИТЕРІЇ (Зв'язок кожної змінної з іншими)
# ==============================================================================
Z = np.linalg.inv(r)
Z_df = pd.DataFrame(Z, index=cols, columns=cols)
print("3. Обернена матриця Z = r^(-1):")
print(Z_df.round(4), "\n")

F_stats = [(Z[i, i] - 1) * (T - N) / (N - 1) for i in range(N)]
F_crit = stats.f.ppf(1 - alpha, dfn=N - 1, dfd=T - N)

print(f"   F-критерії (F_КР = {F_crit:.4f}):")
has_F = False
for col, f_val in zip(cols, F_stats):
    status = "незалежна змінна мультиколінеарна з масивом інших змінних" if f_val > F_crit else "не мультиколінеарна"
    if f_val > F_crit:
        has_F = True
    print(f"   F({col})_СТ = {f_val:.4f} -> {status}")
print()
```

```python
# ==============================================================================
# 4. ЧАСТКОВІ КОЕФІЦІЄНТИ КОРЕЛЯЦІЇ ТА t-КРИТЕРІЇ (Парна мультиколінеарність)
# ==============================================================================
# У підручнику використовується одностороннє критичне значення t(0.05; df)
t_crit = stats.t.ppf(1 - alpha, df=T - N)

print(f"4. Часткові коефіцієнти кореляції та t-критерії (t_КР = {t_crit:.4f}):")
has_t = False
for i in range(N):
    for j in range(i + 1, N):
        r_part = -Z[i, j] / np.sqrt(Z[i, i] * Z[j, j])
        t_stat = abs(r_part) * np.sqrt(T - N) / np.sqrt(1 - r_part**2)
        signif = "частковий коефіцієнт кореляції значимо відрізняється від нуля (є парна колінеарність)" if t_stat > t_crit else "незначимий"
        if t_stat > t_crit:
            has_t = True
        print(f"   r({cols[i]}, {cols[j]} | •) = {r_part:+.4f} | t_СТ = {t_stat:.4f} -> {signif}")
```

```python
# ==============================================================================
# 5. ЗАГАЛЬНИЙ ВИСНОВОК І МЕТОД ГОЛОВНИХ КОМПОНЕНТ (якщо є мультиколінеарність)
# ==============================================================================
print("\n" + "=" * 60)
if has_chi2 or has_F or has_t:
    print("ПІДСУМОК: Хоча б один критерій вказує на наявність мультиколінеарності.")
    print("Переходимо до усунення мультиколінеарності методом головних компонент (МГК):\n")
    
    # Власні числа та власні вектори кореляційної матриці r
    eigvals, eigvecs = np.linalg.eigh(r)
    idx = np.argsort(eigvals)[::-1]  # Сортування за спаданням власних чисел
    eigvals = eigvals[idx]
    eigvecs = eigvecs[:, idx]
    
    for idx_pc, (lam, vec) in enumerate(zip(eigvals, eigvecs.T), start=1):
        terms = " + ".join([f"({v:+.4f})*{c}" for v, c in zip(vec, cols)])
        print(f"   λ_{idx_pc} = {lam:.4f} | z_{idx_pc} = {terms}")
    
    # Критерій інформативності I_p >= 0.8 (80% дисперсії)
    cum_info = np.cumsum(eigvals) / N
    p = int(np.argmax(cum_info >= 0.8) + 1)
    print(f"\n   Критерій інформативності I_{p} = {cum_info[p-1]:.4f} ({cum_info[p-1]*100:.2f}% >= 80%).")
    print(f"   Достатньо залишити перші p = {p} головних компонент.")
    
    # Матриця навантажень A = L^T * Λ^(1/2)
    A = eigvecs[:, :p] * np.sqrt(eigvals[:p])
    A_df = pd.DataFrame(A, index=cols, columns=[f"z_{j+1}" for j in range(p)])
    print("\n   Матриця навантажень A головних компонент на початкові ознаки:")
    print(A_df.round(4))
    
    # Нові некорельовані змінні (головні компоненти) для подальшої регресії
    X_std = (X_norm - X_norm.mean()) / X_norm.std(ddof=1)
    Z_pca = pd.DataFrame(X_std.values @ eigvecs[:, :p], columns=[f"z_{j+1}" for j in range(p)], index=X_norm.index)
else:
    print("ПІДСУМОК: За всіма трьома критеріями мультиколінеарність відсутня.")
```
