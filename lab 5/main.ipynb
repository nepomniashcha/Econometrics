```python
def check_regression_significance(r_squared, T, N, alpha=0.05):
    # 1. Расчет статистики F_ст по формуле
    numerator = r_squared * (T - N)
    denominator = (1 - r_squared) * (N - 1)
    
    # Защита от деления на ноль, если R^2 = 1
    if denominator == 0:
        if numerator > 0:
            F_st = float('inf')
        else:
            F_st = 0 # Случай R^2=0 и T=N (хотя на практике T должно быть > N)
    else:
        F_st = numerator / denominator

    # 2. Определение критического значения F_кр (табличное значение Фишера)
    # Степени свободы: df1 = N - 1, df2 = T - N
    df1 = N - 1
    df2 = T - N
    
    # Находим квантиль распределения Фишера (аналог таблице)
    # Используем 1 - alpha, так как нас интересует правый хвост распределения
    F_kr = stats.f.ppf(1 - alpha, df1, df2)

    # 3. Сравнение F_ст и F_кр
    is_significant = F_st > F_kr

    if is_significant:
        str = f"Рівняння значимо відрізняється від нуля з ймовірністю {(1-alpha)*100:.0f}%."
    else:
        str = f"Рівняння НЕзначимо відрізняється від нуля з ймовірністю {(1-alpha)*100:.0f}%."
    return F_st, F_kr, str

def calculate_metrics(y_true, y_pred, N):
    T = len(y_true)
    y_mean = np.mean(y_true)
    
    # Сума квадратів залишків (RSS) та загальна сума квадратів (TSS)
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - y_mean) ** 2)
    
    # 1. Звичайний R^2
    r_squared = 1 - (ss_res / ss_tot)
    
    # 2. R^2, скоригований за Тейлом (стандартна формула коригування)
    # r_squared_theil = 1 - ((ss_res / (T - N)) / (ss_tot / (T - 1)))
    r_squared_theil = 1 - (1 - r_squared) * ((T - 1) / (T - N))
    
    # 3. R^2, скоригований за Амемією
    r_squared_amemiya = 1 - (1 - r_squared) * ((T + N) / (T - N))

    f_stat, f_krit, str = check_regression_significance(r_squared, T, N, 0.05)
        
    return r_squared, r_squared_theil, r_squared_amemiya, f_stat, f_krit, str
```

```python
y = df['Y1']
results = []

# --- Модель А: Множинна регресія (всі X) ---
beta_A = np.array([8.21702248e+01, -1.95235588e-06, -3.74770557e-01, -1.15007499e+00, 3.26352360e-04, 1.51121836e-02])
# Формуємо матрицю X з одиничним стовпцем для вільного члена
X_matrix_A = np.column_stack((np.ones(len(df)), df['X1'], df['X2'], df['X3'], df['X4'], df['X5']))
y_pred_A = X_matrix_A.dot(beta_A)
results.append(('Модель А (X1-X5)', *calculate_metrics(y, y_pred_A, N=6)))

# --- Моделі Б-Е: Парна регресія ---
models_params = {
    'Модель Б (X1)': (67.88699402392865, -4.978568846458314e-06, 'X1'),
    'Модель В (X2)': (81.49825471956726, -0.6572671284197451, 'X2'),
    'Модель Г (X3)': (80.63353260114002, -1.4803976087863233, 'X3'),
    'Модель Д (X4)': (61.35042901145929, 0.0007279043232142301, 'X4'),
    'Модель Е (X5)': (52.67616578704193, 0.24004969724783892, 'X5'),
}

for model_name, (beta1, beta2, feature) in models_params.items():
    y_pred = beta1 + beta2 * df[feature]
    # N=2 (константа + 1 змінна)
    results.append((model_name, *calculate_metrics(y, y_pred, N=2)))
```

```python
results_df = pd.DataFrame(results, columns=[
    'Модель', 'R^2', 'R^2 (Тейл)', 'R^2 (Амемія)', 'F-статистика', 'F-критерій', 'Значуща (95%)'
])

# Форматування чисел для зручного читання
pd.options.display.float_format = '{:.10f}'.format
print("Результати оцінки моделей:")
print("-" * 80)
print(results_df.to_string(index=False))
```
