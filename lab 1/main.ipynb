```python
# Функція для розрахунку парного коефіцієнта кореляції за формулою
def calc_correlation(x, y):
    n = len(x)
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_xy = np.sum(x * y)
    sum_x_sq = np.sum(x**2)
    sum_y_sq = np.sum(y**2)
    
    numerator = n * sum_xy - sum_x * sum_y
    denominator = np.sqrt((n * sum_x_sq - sum_x**2) * (n * sum_y_sq - sum_y**2))
    
    return numerator / denominator if denominator != 0 else 0

# 2. Задаємо рівень значущості alpha
alpha = 0.2
n = len(df)
print(f"Кількість спостережень (n) = {n}, Рівень значущості (\u03b1) = {alpha}\n")

# Незалежні змінні для аналізу
x_cols = ['X1', 'X2', 'X3', 'X6', 'X8']

for x_col in x_cols:
    # Обчислення коефіцієнта r за власною функцією
    r = calc_correlation(df[x_col], df['Y1'])
    
    # 3. Розрахунок t_CT за формулою алгоритму
    if r**2 == 1:
        t_ct = np.inf
    else:
        t_ct = abs(r) * np.sqrt((n - 2) / (1 - r**2))
    
    # 4. Визначення t_KP (двосторонній критерій Стьюдента)
    t_kp = stats.t.ppf(1 - alpha/2, n - 2)
    
    # 5. Перевірка статистичної гіпотези
    is_significant = t_ct > t_kp
    
    print(f"--- Зв'язок Y1 та {x_col} ---")
    print(f"  Коефіцієнт кореляції (r) = {r:.4f}")
    print(f"  t_CT = {t_ct:.4f}, t_KP = {t_kp:.4f}")
    
    if is_significant:
        print(f"  Висновок: t_CT > t_KP. Гіпотеза H0 відхиляється. Зв'язок значимо відрізняється від нуля з імовірністю {(1-alpha)*100:.0f}%.\n")
    else:
        print("  Висновок: t_CT <= t_KP. Немає підстав відхилити H0. Зв'язок не є статистично значущим.\n")
```
